# Pertemuan 3: Manipulasi Gambar Dasar 



```python
import cv2
import matplotlib.pyplot as plt


def proses_gambar(nama_file, idx, size, crop_area, sudut):
    image = cv2.imread(nama_file)

    resized = cv2.resize(image, size)
    cropped = image[crop_area[0]:crop_area[1], crop_area[2]:crop_area[3]]
    flipped = cv2.flip(image, 1)

    (h, w) = image.shape[:2]
    center = (w/2, h/2)
    M = cv2.getRotationMatrix2D(center, sudut, 1)
    rotated = cv2.warpAffine(image, M, (w, h))

    # Simpan
    cv2.imwrite(f'resize_{idx}.jpg', resized)
    cv2.imwrite(f'crop_{idx}.jpg', cropped)
    cv2.imwrite(f'flip_{idx}.jpg', flipped)
    cv2.imwrite(f'rotate_{idx}.jpg', rotated)

    # Tampilkan perbandingan
    plt.figure(figsize=(10,5))

    plt.subplot(1,5,1)
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title('Asli')

    plt.subplot(1,5,2)
    plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
    plt.title('Resize')

    plt.subplot(1,5,3)
    plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
    plt.title('Crop')

    plt.subplot(1,5,4)
    plt.imshow(cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB))
    plt.title('Flip')

    plt.subplot(1,5,5)
    plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
    plt.title('Rotate')

    plt.show()


# PANGGIL UNTUK 10 GAMBAR


proses_gambar('gambar1.jpg', 1, (400,400), (50,250,50,250), 45)
proses_gambar('gambar2.jpg', 2, (500,300), (30,200,30,200), 60)
proses_gambar('gambar3.jpg', 3, (350,350), (40,220,40,220), 30)
proses_gambar('gambar4.jpg', 4, (600,400), (60,260,60,260), 90)
proses_gambar('gambar5.jpg', 5, (450,450), (50,200,50,200), 120)
proses_gambar('gambar6.jpg', 6, (300,500), (30,180,30,180), 15)
proses_gambar('gambar7.jpg', 7, (500,500), (70,270,70,270), 75)
proses_gambar('gambar8.jpg', 8, (400,300), (40,210,40,210), 135)
proses_gambar('gambar9.jpg', 9, (550,350), (60,230,60,230), 25)
proses_gambar('gambar10.jpg',10,(380,380), (50,220,50,220), 180)
```
